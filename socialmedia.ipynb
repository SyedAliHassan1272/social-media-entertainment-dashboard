```python
important_columns = [
    # Demographics & User Profile
    "Age", "Gender", "Country", "Occupation", "Marital Status",

    # Digital Usage & Screen Time
    "Daily Social Media Time (hrs)", "Daily Entertainment Time (hrs)", 
    "Daily Messaging Time (hrs)", "Daily Video Content Time (hrs)", 
    "Daily Gaming Time (hrs)", "Primary Platform", "Screen Time (hrs)",

    # Internet & Device Usage
    "Device Type", "Internet Speed (Mbps)", "Data Plan Used",

    # Well-being & Mental Health
    "Social Media Fatigue Level (scale 1-10)", "Digital Well-being Awareness",
    "Tech Savviness Level (scale 1-10)", "Average Sleep Time (hrs)", 
    "Sleep Quality (scale 1-10)", "Social Isolation Feeling (scale 1-10)", 
    "Physical Activity Time (hrs)",

    # Financial & Spending Patterns
    "Monthly Income (USD)", "Monthly Expenditure on Entertainment (USD)", "Subscription Platforms",

    # User Engagement & Content Preferences
    "Preferred Content Type", "Primary Social Media Goal", "Preferred Entertainment Platform",
    "Time Spent in Online Communities (hrs)", "News Consumption Time (hrs)"
]

# Filter the DataFrame
df_filtered = df[important_columns]

# Save the filtered dataset for dashboard
df_filtered.to_csv("filtered_dataset.csv", index=False)

# Display first 5 rows
print(df_filtered.head())
```

```python
plt.figure(figsize=(12, 6))
df_filtered.isnull().sum().plot(kind="bar", color="red")
plt.xlabel("Columns")
plt.ylabel("Count of Missing Values")
plt.title("Missing Values in Each Column")
plt.xticks(rotation=90)  # Rotate x labels for readability
plt.show()
```
